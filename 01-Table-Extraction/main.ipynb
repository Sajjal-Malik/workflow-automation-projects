```python
import pandas as pd
import requests
```

```python
# Define the target website
url = "https://en.wikipedia.org/wiki/List_of_largest_banks"
```

```python
# Define a custom header to mimic a standard web browser
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
}
```

```python
# Fetch the content with the updated headers using the requests library
try:
    response = requests.get(url, headers=headers)

    # Check for other HTTP request errors
    response.raise_for_status()

    # Pass the text content of the response into pandas
    tables = pd.read_html(response.text)

    # Output details to the terminal
    print(f"Total tables found: {len(tables)}\n")

    # Select and display the first table
    df = tables[0]
    print("--- First Table Preview ---")
    print(df.head())

except requests.exceptions.RequestException as e:
    print(f"An error occurred while fetching the page: {e}")
```
